# matplotlib 基本画图流程
1、导入库
2、准备数据
3、调用绘图函数
4、添加标签、标题、图例
5、显示图像

````python
# 1、导入库
import matplotlib.pyplot as plt
import numpy as np
"""
所有matplotlib的骨架：
plt.figure(figsize=(8, 4))

plt.plot(...)

plt.xlabel(...)
plt.ylabel(...)
plt.title(...)
plt.legend()
plt.grid(True)

plt.show()

重新拿出一个画布来做图
plt.figure()   # 拿出一张白纸
plt.plot()     # 在纸上画东西
plt.show()     # 把纸展示给别人看
"""
````

````python
# ===== plot() 画线 函数详解 =====
# 用法：plt.plot(x, y, fmt, label, color, linewidth, ...)

# e.g.1 最简单的用法
x = np.linspace(0, 2*np.pi, 100)  # np.linspace(start, stop, num)在指定的间隔内返回均匀间隔的数字。这里生成了从0到2π之间的100个点。
y = np.sin(x)

plt.figure(figsize=(10, 5)) #创建一个新的画布（Figure）设置图像大小，单位为英寸 (宽, 高)，默认值为 (6.4, 4.8)
plt.plot(x, y, label='sin(x)') 
"""
plot()函数用法：
plt.plot(
    x,              如果传入一个一维数组，则默认y为该数组的索引；然后x自动生成[0, 1, 2, ...]。
    y,              如果传入两个一维数组，则x和y分别表示横轴和纵轴的数据。
    color='red',    # 颜色：'r'(红) 'g'(绿) 'b'(蓝) 'k'(黑) 'y'(黄) 'm'(洋红) 'c'(青) 'w'(白)
    linestyle='--', # 线型：'-'(实线) '--'(虚线) '-.'(点划线) ':'(点线)
    linewidth=2,    # 线宽，默认1.5
    marker='o',     # 标记：'o'(圆) '*'(星) '+'(加) 'x'(叉) 's'(方) '^'(三角)
    label='sin(x)'  # 图例标签

)
"""
plt.xlabel('x') # 设置x轴标签
plt.ylabel('y') # 设置y轴标签
plt.title('sin 函数') # 设置图像标题
plt.legend() # 显示图例只有调用 plt.legend() 才会显示图例。
plt.grid(True) # 显示网格
plt.show() # 显示图像



# e.g.2 ：红色虚线 + 圆形标记
plt.plot(x, y, '--ro', label='sin(x)')
plt.legend()
plt.show()



# e.g.3 ：多条线
plt.figure(figsize=(10, 5))
plt.plot(x, np.sin(x), '-r', label='sin(x)', linewidth=2)
plt.plot(x, np.cos(x), '--b', label='cos(x)', linewidth=2)
plt.plot(x, np.tan(x), ':g', label='tan(x)', linewidth=3.5)
plt.xlabel('x')
plt.ylabel('y')
plt.title('三角函数比较')
plt.legend()
plt.grid(True)
plt.ylim(-5, 5)  # 限制 y 轴范围
plt.show()

# 关键参数
# color：颜色，如 'red', '#FF0000'
# linewidth (lw)：线宽，默认 1.5
# linestyle (ls)：线型
# marker：标记点形状
# markersize (ms)：标记大小
# alpha：透明度 (0-1)
# label：图例标签
````

````python
# ===== scatter() 画点 函数详解 =====
# 用法：plt.scatter(x, y, s=size, c=color, marker='o', alpha=0.5, ...)

# 基础散点图
np.random.seed(42) # seed = 随机数种子 42 就是随机数的起点
x = np.random.randn(50)  # randn = random normal 从正态分布随机抽取50个数字
y = np.random.randn(50)

plt.figure(figsize=(8, 6))
plt.scatter(x, y, c='blue', s=50, marker='o')
"""
plt.scatter(
    x,          # 横坐标
    y,          # 纵坐标
    c='blue',   # 颜色(color)   # c：颜色，可以是单一颜色或数组（用于映射到 cmap）
    s=50,       # 点大小(size). # s：点大小（标量或数组），默认 36
    marker='o', # 点形状        # marker：点形状，默认 'o'，可选 's'(方), '^'(三角), '*'(星) , 'x' 叉号,'v'   # 下三角
    alpha=1.0,  # 透明度.       # 0 = 完全透明; 1 = 完全不透明
    label=''    # 图例名称
    edgecolor   # 边框颜色
    cmap=" "    # 颜色映射      # 'viridis', 'hot', 'cool', 'gray'——相当于调色板，当c是数组时要把c的数值变成颜色
    linewidth=  # 边框宽度
)
"""
plt.xlabel('x')
plt.ylabel('y')
plt.title('Scatter plot')
plt.grid(True)
plt.show()

# 带颜色分布的散点图
x = np.random.randn(100)
y = np.random.randn(100)
z = x + y  # 第三维数据

plt.figure(figsize=(8, 6))
scatter = plt.scatter(x, y, c=z, s=100, cmap='plasma', alpha=0.6)
plt.colorbar(scatter, label='x+y')  # 添加颜色条
plt.xlabel('x')
plt.ylabel('y')
plt.title('Scatter plot with color gradient')
plt.show()

# 不同大小的散点
x = np.random.rand(50)
y = np.random.rand(50)
sizes = np.random.rand(50) * 200  # 大小随机

plt.figure(figsize=(8, 6))
plt.scatter(x, y, s=sizes, alpha=0.5, c=sizes, cmap='hot')
plt.colorbar(label='Size')
plt.title('Scatter with different sizes')
plt.show()

"""
cmap         风格

viridis	紫→蓝→绿→黄（默认推荐）

plasma	紫→红→黄，更鲜艳

inferno 黑→红→黄，对比强

coolwarm 蓝↔红，适合“正负”数据

gray	黑白灰

"""
````

````python
# ===== bar() 函数详解 =====
# 用法：plt.bar(x, height, width, color, label, ...)

# e.g.1 基础柱状图
# 相当于x，y
categories = ['A', 'B', 'C', 'D', 'E']
values = [5, 7, 3, 8, 4]

plt.figure(figsize=(8, 5))
plt.bar(categories, values, color='skyblue', edgecolor='navy')
plt.xlabel('Category')
plt.ylabel('Value')
plt.title('Bar chart')
plt.grid(axis='y', alpha=0.3) # 打开网格线，只显示 y 方向网格，并且设置透明度为 0.3。
plt.show()
"""
plt.bar(
    x,           # 柱子的位置
    height,      # 柱子的高度
    width=0.8,   # 柱子的宽度
    color='blue',
    label='Data'
)
"""


# e.g.2 多组数据（分组柱状图）
categories = ['A', 'B', 'C']
group1 = [5, 7, 3]
group2 = [6, 4, 8]

x = np.arange(len(categories))
width = 0.35  # 柱子宽度

plt.figure(figsize=(8, 5))
plt.bar(x - width/2, group1, width, label='Group1', color='skyblue')
plt.bar(x + width/2, group2, width, label='Group2', color='lightcoral')
plt.xlabel('Category')
plt.ylabel('Value')
plt.xticks(x, categories)  # x轴标签
plt.legend()
plt.show()



# e.g.3 水平柱状图（barh）
categories = ['项目A', '项目B', '项目C', '项目D']
values = [23, 17, 35, 29]

plt.figure(figsize=(8, 5))
plt.barh(categories, values, color='lightgreen')  # plt.barh() 水平柱状图
plt.xlabel('Value')
plt.title('Horizontal bar chart')
plt.grid(axis='x', alpha=0.3)
plt.show()



# e.g.4 堆积柱状图
categories = ['Q1', 'Q2', 'Q3', 'Q4']
data1 = [20, 25, 30, 28]
data2 = [15, 18, 22, 20]
data3 = [10, 12, 15, 14]

plt.figure(figsize=(8, 5))
plt.bar(categories, data1, label='Product A', color='skyblue')
plt.bar(categories, data2, bottom=data1, label='Product B', color='lightcoral')
# bottom=data1 ———— Product B 不要从 0 开始画，而是从 Product A 顶部开始画。
bottom = [data1[i] + data2[i] for i in range(len(data1))]
plt.bar(categories, data3, bottom=bottom, label='Product C', color='lightgreen')
plt.ylabel('Sales')
plt.title('Stacked bar chart')
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.show()

# 关键参数
# x：类别或位置
# height：柱子高度（数值）
# width：柱子宽度，默认 0.8
# color：颜色
# edgecolor：边框颜色
# label：图例标签
# alpha：透明度
# bottom：柱子底部起点（用于堆积）
````

````python
# ===== imshow() 函数详解 =====
# 用法：plt.imshow(X, cmap, vmin, vmax, aspect, ...)

# e.g.1 基础热力图（2D 矩阵可视化）
matrix = np.random.rand(10, 10) # np.random.rand() 从 0~1 之间均匀随机生成数字. np.random.rand(10, 10) 生成一个 10 行 10 列的随机矩阵。

plt.figure(figsize=(8, 6))
plt.imshow(matrix, cmap='viridis',interpolation='bilinear')
plt.colorbar(label='Value')
plt.title('Heatmap - viridis')
plt.xlabel('Column')
plt.ylabel('Row')
plt.show()
"""
imshow() = 把一个二维矩阵显示成图片。
lt.imshow(
    X,
    cmap=None,   # 数字 → 什么颜色
    vmin=None,   # 颜色最小值。vmin=0；表示0以下全部按最深颜色显示
    vmax=None,   # 颜色最大值。vmax=1；1以上全部按最亮颜色显示
    interpolation=None,  # interpolation 插值方式。控制像素放大后是否平滑(nearest像素块明显；bilinear过渡更平滑)
    aspect=None  # 控制宽高比例（equal 每个像素保持正方形；auto 自动拉伸。）
    alpha        # 透明度
)


"""



# e.g.2 不同的色彩映射（cmap）
cmaps = ['gray', 'hot', 'cool', 'jet', 'RdYlBu', 'twilight']
matrix = np.random.rand(10, 10)

fig, axes = plt.subplots(2, 3, figsize=(12, 8))
# subplots(2,3)一次性创建多个子图两行三列。fig整个大画布
# axes 6个小坐标轴a
# axes =
#[
# [ax00, ax01, ax02],
# [ax10, ax11, ax12]
#]
for idx, cmap_name in enumerate(cmaps):
    ax = axes[idx // 3, idx % 3]
    im = ax.imshow(matrix, cmap=cmap_name)
    ax.set_title(f'cmap: {cmap_name}')
    plt.colorbar(im, ax=ax)
plt.tight_layout()
plt.show()

# enumerate 返回索引 + 元素
# 0//3 = 0；1//3 = 0；2//3 = 0；3//3 = 1……得到行号，% 得到列号
# 不是plt.imshow() 而是ax.imshow 你要说明白画在哪里
# colorbar(im, ax=ax)； im 告诉colorbar使用哪个颜色映射；ax=ax 挂在哪个子图旁边
# plt.tight_layout() 自动调整间距; 不加就会挤在一起




# e.g.3 图像像素可视化
from scipy import misc
# 如果没有 scipy，可以用随机图像
image = np.random.randint(0, 256, (50, 50, 3), dtype=np.uint8)
# 在 0 ≤ x < 256 （图片经常用uint8存储范围就是0～255）中 随机生成 50行50列，3个通道的整数 = 50×50个像素 每个像素有3个数字（50×50个RGB像素）
plt.figure(figsize=(8, 6))
plt.imshow(image) # 把50×50×3张量解释成RGB图片
# 对于shape=(50,50)二维矩阵显示热力图
# 对于shape=(50,50,3)的RGB图像显示彩色图片。
plt.title('Random image')
plt.axis('off')  # 关闭坐标轴
plt.show()




#e.g.4  灰度图像
gray_image = np.random.randint(0, 256, (50, 50), dtype=np.uint8)

plt.figure(figsize=(8, 6))
plt.imshow(gray_image, cmap='gray')
plt.colorbar(label='Intensity')
plt.title('Grayscale image')
plt.show()




# e.g.5 热力图（相关矩阵）
data = np.random.randn(5, 5)
correlation = np.corrcoef(data)  # 计算相关系数 范围-1 ~ 1（-1完全负相关，1完全正相关，0无关）

plt.figure(figsize=(8, 6))
im = plt.imshow(correlation, cmap='RdYlBu', vmin=-1, vmax=1)
# RdYlBu red：1 yellow：0 blue：-1
plt.colorbar(im, label='Correlation')
plt.title('Correlation heatmap')
# 添加文字注释
for i in range(correlation.shape[0]):
    for j in range(correlation.shape[1]):
        plt.text(j, i, f'{correlation[i, j]:.2f}', 
                ha='center', va='center', color='black', fontsize=10)
                # ha = horizontal alignment 水平居中
                # va = vertical alignment 竖直居中
                # fontsize=10 字体大小。
plt.show()

# 关键参数
# X：输入数据（2D 数组）
# cmap：色彩映射，常见有
#   - 'gray', 'binary'：灰度
#   - 'hot', 'cool', 'spring', 'summer', 'autumn', 'winter'：热/冷色系
#   - 'viridis', 'plasma', 'magma'：感知均匀
#   - 'jet', 'RdYlBu', 'RdBu'：散度
#   - 'twilight'：圆形
# vmin, vmax：值范围，用于映射颜色
# aspect：纵横比，'auto', 'equal'
# interpolation：插值方式，'nearest', 'bilinear', 'bicubic'
# origin：'upper' 或 'lower'，决定坐标系方向
````

# 进阶：子图与布局

````python
# ===== 方法1：plt.subplot() =====
x = np.linspace(0, 2*np.pi, 100)

plt.figure(figsize=(12, 8))

# 第1个图：2行2列，第1个位置
plt.subplot(2, 2, 1)
plt.plot(x, np.sin(x), 'r-')
plt.title('sin(x)')
plt.grid(True)

# 第2个图：2行2列，第2个位置
plt.subplot(2, 2, 2)
plt.plot(x, np.cos(x), 'g-')
plt.title('cos(x)')
plt.grid(True)

# 第3个图：2行2列，第3个位置
plt.subplot(2, 2, 3)
plt.plot(x, np.sin(x)**2, 'b-')
plt.title('sin²(x)')
plt.grid(True)

# 第4个图：2行2列，第4个位置
plt.subplot(2, 2, 4)
plt.plot(x, x, 'k-')
plt.title('y=x')
plt.grid(True)

plt.tight_layout()  # 自动调整间距
plt.show()

# ===== 方法2：fig, ax = plt.subplots() （推荐） =====
x = np.linspace(0, 2*np.pi, 100)
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# axes 是一个 2x2 的数组
axes[0, 0].plot(x, np.sin(x), 'r-')
axes[0, 0].set_title('sin(x)')
axes[0, 0].grid(True)

axes[0, 1].plot(x, np.cos(x), 'g-')
axes[0, 1].set_title('cos(x)')
axes[0, 1].grid(True)

axes[1, 0].plot(x, np.sin(x)**2, 'b-')
axes[1, 0].set_title('sin²(x)')
axes[1, 0].grid(True)

axes[1, 1].plot(x, x, 'k-')
axes[1, 1].set_title('y=x')
axes[1, 1].grid(True)

plt.tight_layout()
plt.show()

# ===== GridSpec 精细控制 =====
fig = plt.figure(figsize=(10, 8))

# 创建 3x3 网格，第1个图占 1x2
ax1 = plt.subplot2grid((3, 3), (0, 0), colspan=2)
ax1.plot(x, np.sin(x))
ax1.set_title('Large subplot')

# 右上角
ax2 = plt.subplot2grid((3, 3), (0, 2))
ax2.scatter(np.random.rand(10), np.random.rand(10))
ax2.set_title('Scatter')

# 下方占满宽度
ax3 = plt.subplot2grid((3, 3), (1, 0), colspan=3, rowspan=2)
ax3.imshow(np.random.rand(10, 10), cmap='viridis')
ax3.set_title('Heatmap')

plt.tight_layout()
plt.show()
````

# 常用设置：轴、标签、图例

````python
# ===== 坐标轴设置 =====
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.figure(figsize=(10, 6))
plt.plot(x, y)

# 设置轴范围
plt.xlim(0, 10)
plt.ylim(-1.5, 1.5)

# 设置轴标签
plt.xlabel('Time (s)', fontsize=12, fontweight='bold')
plt.ylabel('Amplitude', fontsize=12, fontweight='bold')
plt.title('Sine Wave', fontsize=14, fontweight='bold')

# 设置刻度
plt.xticks(np.arange(0, 11, 2))  # 自定义 x 轴刻度
plt.yticks([-1, -0.5, 0, 0.5, 1])  # 自定义 y 轴刻度

# 旋转刻度标签
ax = plt.gca()
ax.tick_params(axis='both', which='major', labelsize=10)

# 添加网格
plt.grid(True, alpha=0.3, linestyle='--')

# 设置坐标轴可见性
ax.spines['right'].set_visible(False)  # 隐藏右边框
ax.spines['top'].set_visible(False)    # 隐藏上边框

plt.show()

# ===== 图例（Legend）设置 =====
x = np.linspace(0, 2*np.pi, 100)

plt.figure(figsize=(10, 6))
plt.plot(x, np.sin(x), label='sin(x)', linewidth=2)
plt.plot(x, np.cos(x), label='cos(x)', linewidth=2)
plt.plot(x, np.tan(x), label='tan(x)', linewidth=2)

# 图例位置：'upper left', 'upper right', 'lower left', 'lower right', 'center'
# 'best' 会自动选择最佳位置（默认）
plt.legend(loc='upper right', fontsize=12, framealpha=0.9)

plt.xlabel('x')
plt.ylabel('y')
plt.title('Trigonometric functions')
plt.ylim(-5, 5)
plt.grid(True, alpha=0.3)
plt.show()

# ===== 文字标注 =====
x = np.linspace(0, 2*np.pi, 100)
y = np.sin(x)

plt.figure(figsize=(10, 6))
plt.plot(x, y, 'b-', linewidth=2)

# 添加文字
plt.text(np.pi/2, 1, '最大值', fontsize=12, ha='center', 
         bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.5))

# 添加箭头注释
plt.annotate('最大值点', xy=(np.pi/2, 1), xytext=(np.pi/2 + 1, 1.3),
            arrowprops=dict(arrowstyle='->', color='red', lw=2),
            fontsize=11, color='red')

plt.xlabel('x')
plt.ylabel('sin(x)')
plt.title('Annotated plot')
plt.grid(True, alpha=0.3)
plt.show()

# ===== 常用函数总结 =====
# plt.xlabel(label, fontsize, fontweight)：x 轴标签
# plt.ylabel(label, fontsize, fontweight)：y 轴标签
# plt.title(label, fontsize, fontweight)：标题
# plt.xlim(xmin, xmax)：x 轴范围
# plt.ylim(ymin, ymax)：y 轴范围
# plt.xticks(ticks, labels, rotation)：x 轴刻度
# plt.yticks(ticks, labels)：y 轴刻度
# plt.legend(loc, fontsize, framealpha)：图例
# plt.grid(True/False, alpha, linestyle)：网格
# plt.text(x, y, text, fontsize, ha, va)：添加文字
# plt.annotate(text, xy, xytext, arrowprops)：添加注释
# plt.axis('off')：关闭坐标轴
# plt.tight_layout()：自动调整布局
````

# 其他常用操作

````python
# ===== 保存图像 =====
# 创建一个简单的图
x = np.linspace(0, 2*np.pi, 100)
plt.figure(figsize=(8, 5))
plt.plot(x, np.sin(x), label='sin(x)')
plt.legend()
plt.title('My plot')

# 保存为不同格式
# plt.savefig('my_plot.png', dpi=300, bbox_inches='tight')  # PNG 格式
# plt.savefig('my_plot.pdf', dpi=300, bbox_inches='tight')  # PDF 格式
# plt.savefig('my_plot.jpg', dpi=300, bbox_inches='tight')  # JPG 格式

# savefig 参数：
# dpi：分辨率（点数/英寸），通常 100-300
# bbox_inches='tight'：自动裁剪空白边界
# format：文件格式，'png', 'pdf', 'jpg'
# facecolor：背景颜色
# edgecolor：边框颜色
# transparent：背景透明

plt.show()

# ===== 面向对象风格（推荐，用于复杂图形）=====
# 创建 figure 和 axes 对象
fig, ax = plt.subplots(figsize=(10, 6))

x = np.linspace(0, 2*np.pi, 100)
ax.plot(x, np.sin(x), 'r-', label='sin(x)', linewidth=2)
ax.plot(x, np.cos(x), 'b--', label='cos(x)', linewidth=2)

# 用 ax 对象的方法设置各种属性
ax.set_xlabel('x', fontsize=12)
ax.set_ylabel('y', fontsize=12)
ax.set_title('Sine and Cosine', fontsize=14, fontweight='bold')
ax.set_xlim(0, 2*np.pi)
ax.set_ylim(-1.5, 1.5)
ax.legend(loc='upper right', fontsize=11)
ax.grid(True, alpha=0.3)

# 添加文字
ax.text(np.pi, 0, 'Origin', fontsize=10, ha='center')

plt.tight_layout()
plt.show()

# ===== 对数坐标 =====
x = np.linspace(1, 100, 50)
y = x**2

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# 普通坐标
axes[0].plot(x, y, 'b-', linewidth=2)
axes[0].set_title('Linear scale')
axes[0].set_xlabel('x')
axes[0].set_ylabel('y')
axes[0].grid(True)

# 对数坐标
axes[1].loglog(x, y, 'r-', linewidth=2)  # 两个轴都是对数
axes[1].set_title('Log-log scale')
axes[1].set_xlabel('x (log scale)')
axes[1].set_ylabel('y (log scale)')
axes[1].grid(True, which='both', alpha=0.3)

plt.tight_layout()
plt.show()

# ===== 其他有用的绘图函数 =====
# plt.hist(data, bins=10)：直方图
# plt.box_plot(data)：箱线图
# plt.contour(X, Y, Z)：等高线
# plt.quiver(X, Y, U, V)：向量场
# plt.fill_between(x, y1, y2)：填充区域
# plt.hexbin(x, y, gridsize=10)：六边形热力图
# plt.pie(data, labels)：饼图

# 直方图示例
data = np.random.randn(1000)
plt.figure(figsize=(10, 5))
plt.hist(data, bins=50, color='skyblue', edgecolor='navy', alpha=0.7)
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram of normal distribution')
plt.grid(True, alpha=0.3, axis='y')
plt.show()
````

# 知识点总结

## 四大绘图函数
| 函数 | 用途 | 参数 |
|------|------|------|
| `plot()` | 线图 | fmt, linewidth, marker, alpha, label |
| `scatter()` | 散点图 | s(大小), c(颜色), cmap, marker, alpha |
| `bar()` | 柱状图 | height, width, color, label |
| `imshow()` | 热力图/矩阵 | cmap, vmin, vmax, aspect, interpolation |

## 坐标轴设置
- `xlim(xmin, xmax)` / `ylim(ymin, ymax)`：设置范围
- `xticks() / yticks()`：设置刻度
- `xlabel() / ylabel()`：设置标签
- `title()`：设置标题
- `grid()`：显示网格
- `legend()`：显示图例

## 布局
- `subplot(m, n, i)`：简单布局（m行n列第i个）
- `subplots(m, n)`：面向对象风格（推荐）
- `tight_layout()`：自动调整间距

## 常用参数
- `figsize=(width, height)`：图形大小
- `alpha`：透明度（0-1）
- `color`：颜色（'r', 'g', 'b' 或 '#FF0000'）
- `linewidth (lw)`：线宽
- `linestyle (ls)`：线型（'-', '--', '-.'）
- `marker`：标记点（'o', 's', '^'）
- `label`：用于图例
- `cmap`：色彩映射

## 面向对象 vs 函数式
```python
# 函数式（简单）
plt.plot(x, y)
plt.show()

# 面向对象（灵活）
fig, ax = plt.subplots()
ax.plot(x, y)
```
