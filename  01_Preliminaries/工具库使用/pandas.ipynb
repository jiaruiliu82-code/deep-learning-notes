# pandas 完整教程

pandas 是 Python 中最流行的数据处理库，用于表格数据的读取、清洗、转换、分析。

## pandas 的两个核心数据结构
1. **Series** — 一维数据（列）
2. **DataFrame** — 二维数据（表格）

```python
# 导入库
import pandas as pd
import numpy as np

# ===== 1. Series —— 一维数据结构 =====
# series = 带标签的一维数组
# 方式1：从列表创建
s1 = pd.Series([1, 2, 3, 4])
print("Series 示例1:")
print(s1)
print("\n")

# 方式2：指定索引（index）
s2 = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
print("Series 示例2（有自定义索引）:")
print(s2)
print("\n")

# 方式3：从字典创建
s3 = pd.Series({'北京': 2100, '上海': 2400, '深圳': 1500})
print("Series 示例3（从字典）:")
print(s3)
print("\n")

# Series 属性
print(f"Series 的值: {s2.values}")
print(f"Series 的索引: {s2.index.tolist()}")
print(f"Series 的长度: {len(s2)}")
print(f"Series 的数据类型: {s2.dtype}")
print("\n")

# 访问 Series 元素
print(f"按位置访问 s2[0]: {s2[0]}")
print(f"按索引访问 s2['a']: {s2['a']}")
print("\n")
```

# DataFrame —— 二维数据结构

```python
# ===== 2. DataFrame 创建方式 =====
# 核心：多列series组成的表格


# 方式1：从字典创建
df1 = pd.DataFrame({
    '姓名': ['小明', '小红', '小李'],
    '年龄': [25, 23, 24],
    '城市': ['北京', '上海', '深圳']
})
print("DataFrame 示例1（从字典）:")
print(df1)
print("\n")
# 键会变成列名；值会变成这一列的数据



# 方式2：从列表创建，指定列名
df2 = pd.DataFrame(
    [[1, 2, 3], [4, 5, 6], [7, 8, 9]],
    columns=['A', 'B', 'C']
)
print("DataFrame 示例2（从列表）:")
print(df2)
print("\n")

# 方式3：从 Series 字典创建 其实也就是方式1变成可以改 index 了
df3 = pd.DataFrame({
    '销售额': pd.Series([100, 200, 150], index=['Q1', 'Q2', 'Q3']),
    '利润率': pd.Series([0.1, 0.15, 0.12], index=['Q1', 'Q2', 'Q3'])
})
print("DataFrame 示例3（从 Series）:")
print(df3)
print("\n")

# DataFrame 基本属性
print(f"形状: {df1.shape}")  # (行数, 列数)
print(f"列名: {df1.columns.tolist()}")
print(f"行索引: {df1.index.tolist()}")
print(f"数据类型:\n{df1.dtypes}")
print("\n")

# DataFrame 基本信息
print("DataFrame 信息:")
df1.info()
print("\n")

# 数据摘要统计
print("统计信息:")
print(df1.describe())
print("\n")
```

# 数据索引与选择

```python
# ===== 5. 数据索引与选择 =====

df = pd.DataFrame({
    '姓名': ['张三', '李四', '王五', '赵六', '孙七'],
    '年龄': [25, 30, 28, 22, 35],
    '工资': [8000, 12000, 10000, 7000, 15000],
    '部门': ['IT', 'HR', 'IT', '销售', 'IT']
})

print("原始 DataFrame:")
print(df)
print("\n")

# ===== 按位置索引（iloc） =====
print("按位置索引 iloc:")
print("第 0 行:", df.iloc[0].tolist())
print("第 1-3 行:", "\n", df.iloc[1:3])
print("第 0-2 列:", "\n", df.iloc[:, 0:2])
print("第 1 行，第 2 列的值:", df.iloc[1, 2])
print("\n")

"""
df.iloc[行位置, 列位置]
"""

# ===== 按标签索引（loc） =====
print("按标签索引 loc:")
print("第 0 行:", df.loc[0].tolist())
print("第 0-2 行:", "\n", df.loc[0:2])
print("第 1 行，'工资' 列:", df.loc[1, '工资'])
print("\n")

# ===== 条件选择 =====
print("条件选择 - 工资 > 9000:")
condition = df['工资'] > 9000
print(df[condition])
print("\n")

print("多条件选择 - 年龄 > 25 且 部门 == 'IT':")
condition = (df['年龄'] > 25) & (df['部门'] == 'IT')
print(df[condition])
print("\n")

# ===== isin() 方法 =====
print("isin() 方法 - 部门为 IT 或 HR:")
condition = df['部门'].isin(['IT', 'HR'])
print(df[condition])
print("\n")
"""
某个值是否属于给定集合 类似于python中 x in [0,1,2,3]
df['部门'].isin(['IT', 'HR'])  相当于 部门 == IT 或 部门 == HR
"""
# ===== query() 方法 =====
print("query() 方法 - 工资 > 9000:")
print(df.query('工资 > 9000'))
print("\n")

"""
把筛选条件写成字符串。
普通：df[df['工资'] > 9000]
query：df.query('工资 > 9000')
"""
```

# 数据清洗与处理

```python
# ===== 6. 数据清洗与处理 =====

# 创建有缺失值的 DataFrame
df_dirty = pd.DataFrame({
    '姓名': ['张三', '李四', None, '赵六', '孙七'],
    '年龄': [25, np.nan, 28, 22, 35],
    '工资': [8000, 12000, 10000, None, 15000]
})

print("原始有缺失值的 DataFrame:")
print(df_dirty)
print("\n")

# ===== 处理缺失值 =====
print("检查缺失值:")  # isnull() 会把 DataFrame 里每个元素变成 True / False True = 这个位置是缺失值（NaN）；False = 正常值
print(df_dirty.isnull())
print("\n")

print("缺失值数量:")
print(df_dirty.isnull().sum())
print("\n")

# 删除含有缺失值的行
print("删除含有缺失值的行（dropna）:")
print(df_dirty.dropna())
print("\n")

# 填充缺失值
print("用 0 填充缺失值（fillna）:")
print(df_dirty.fillna(0))
print("\n")

print("用前一行值填充（forward fill）:")
print(df_dirty.fillna(method='ffill'))
print("\n")

# ===== 重复值处理 =====
df_dup = pd.DataFrame({
    'A': [1, 1, 2, 2, 3],
    'B': ['a', 'a', 'b', 'b', 'c']
})

print("有重复行的 DataFrame:")
print(df_dup)
print("\n")

print("检查重复行:")
print(df_dup.duplicated())
print("\n")

print("删除重复行 (drop_duplicates):")
print(df_dup.drop_duplicates())
print("\n")

# ===== 数据类型转换 =====
df_type = pd.DataFrame({
    'A': ['1', '2', '3'],
    'B': ['10.5', '20.3', '15.8']
})

print("原始数据类型:")
print(df_type.dtypes)
print("\n")

# 转换数据类型
df_type['A'] = df_type['A'].astype(int)
df_type['B'] = df_type['B'].astype(float)

print("转换后的数据类型:")
print(df_type.dtypes)
print("\n")

# ===== 字符串操作 =====
df_str = pd.DataFrame({'名字': ['张 三', ' 李四 ', '王-五']})

print("原始字符串:")
print(df_str)
print("\n")

print("去除空格 (str.strip):")
print(df_str['名字'].str.strip())
print("\n")

print("大小写转换 (str.upper):")
print(df_str['名字'].str.upper())
print("\n")

print("替换 (str.replace):")
print(df_str['名字'].str.replace('-', '_'))
print("\n")
```

# 数据分组与聚合

```python
# ===== 7. 数据分组与聚合 =====
"""
在回答：按部门统计数据」

比如：

IT部门总工资是多少？
IT部门平均工资是多少？
每个部门有几个人？
每个人比部门平均工资高多少？

"""

df = pd.DataFrame({
    '部门': ['IT', 'IT', 'HR', 'HR', 'IT', '销售'],
    '姓名': ['张三', '李四', '王五', '赵六', '孙七', '周八'],
    '工资': [8000, 12000, 9000, 8500, 10000, 7500],
    '绩效': [85, 90, 80, 85, 92, 75]
})

print("原始 DataFrame:")
print(df)
print("\n")

# ===== 按部门分组求和 =====
print("按部门分组 - 工资总和:")
print(df.groupby('部门')['工资'].sum()) 
# 按照"部门"这一列进行分组，变成IT:张三、李四、孙七 ； HR:王五、赵六 ；销售: 周八
# 只取工资列进行求和
print("\n")

# 多个聚合函数
print("按部门分组 - 多个统计:")
print(df.groupby('部门')['工资'].agg(['sum', 'mean', 'count']))
print("\n")

# 多列分组
print("按部门分组 - 多列求和:")
print(df.groupby('部门')[['工资', '绩效']].sum())
print("\n")

# 自定义聚合函数
print("按部门分组 - 自定义函数（最大值 - 最小值）:")
print(df.groupby('部门')['工资'].agg(lambda x: x.max() - x.min()))
print("\n")

# 多个聚合函数（不同列用不同函数）
print("按部门分组 - 不同列用不同聚合函数:")
agg_dict = {'工资': 'sum', '绩效': 'mean'}
print(df.groupby('部门').agg(agg_dict))
print("\n")

# 组内转换（transform）
print("每个人的工资与部门平均工资的差:")
df['工资偏差'] = df['工资'] - df.groupby('部门')['工资'].transform('mean')
print(df[['姓名', '部门', '工资', '工资偏差']])
print("\n")
# 但 transform 不会减少行数，把部门平均值扩展回原来的每一行

# 过滤分组（filter）
print("只保留部门平均工资 > 8500 的员工:")
print(df.groupby('部门').filter(lambda x: x['工资'].mean() > 8500))
print("\n")
```

# 数据合并与拼接

```python
# ===== 8. 数据合并与拼接 =====

# ===== concat() - 拼接 =====
df1 = pd.DataFrame({
    '姓名': ['张三', '李四'],
    '工资': [8000, 12000]
})

df2 = pd.DataFrame({
    '姓名': ['王五', '赵六'],
    '工资': [10000, 9000]
})

print("df1:")
print(df1)
print("\ndf2:")
print(df2)
print("\n")

# 行拼接（纵向）
print("行拼接（concat，axis=0）:")
print(pd.concat([df1, df2], ignore_index=True))
print("\n")

# 列拼接（横向）
df3 = pd.DataFrame({
    '年龄': [25, 30, 28, 22]
})
print("列拼接（concat，axis=1）:")
print(pd.concat([pd.concat([df1, df2], ignore_index=True), df3], axis=1))
print("\n")

# ===== merge() - 左右合并 =====
df_employee = pd.DataFrame({
    '员工ID': [1, 2, 3, 4],
    '姓名': ['张三', '李四', '王五', '赵六'],
    '部门ID': [10, 20, 10, 30]
})

df_dept = pd.DataFrame({
    '部门ID': [10, 20, 30],
    '部门名': ['IT', 'HR', '销售']
})

print("员工表:")
print(df_employee)
print("\n部门表:")
print(df_dept)
print("\n")

# 内连接（inner）
print("内连接（merge，how='inner'）:")
print(pd.merge(df_employee, df_dept, on='部门ID', how='inner'))
print("\n")

# 左连接（left）
print("左连接（merge，how='left'）:")
print(pd.merge(df_employee, df_dept, on='部门ID', how='left'))
print("\n")

# 右连接（right）
print("右连接（merge，how='right'）:")
print(pd.merge(df_employee, df_dept, on='部门ID', how='right'))
print("\n")

# 外连接（outer）
print("外连接（merge，how='outer'）:")
print(pd.merge(df_employee, df_dept, on='部门ID', how='outer'))
print("\n")

# ===== join() - 按索引合并 =====
df_a = pd.DataFrame({
    'A': [1, 2, 3],
    'B': [4, 5, 6]
}, index=['a', 'b', 'c'])

df_b = pd.DataFrame({
    'C': [7, 8, 9],
    'D': [10, 11, 12]
}, index=['a', 'b', 'c'])

print("df_a:")
print(df_a)
print("\ndf_b:")
print(df_b)
print("\n")

print("按索引 join:")
print(df_a.join(df_b))
print("\n")
```

# 数据转换与排序

```python
# ===== 9. 数据转换与排序 =====

df = pd.DataFrame({
    '姓名': ['张三', '李四', '王五'],
    '工资': [8000, 12000, 10000],
    '评分': [85, 90, 78]
})

print("原始 DataFrame:")
print(df)
print("\n")

# ===== apply() - 按行或按列应用函数 =====
print("apply() - 对每一行应用自定义函数:")
def raise_salary(row):
    if row['评分'] >= 90:
        return row['工资'] * 1.2
    else:
        return row['工资']

df['调整后工资'] = df.apply(raise_salary, axis=1)
print(df)
print("\n")

# 对列应用函数
print("apply() - 对工资列乘以 1.1:")
print(df['工资'].apply(lambda x: x * 1.1))
print("\n")

# ===== map() - 映射值 =====
grade_map = {85: 'B', 90: 'A', 78: 'C'}
print("map() - 评分映射到等级:")
print(df['评分'].map(grade_map))
print("\n")

# ===== 排序 =====
print("按工资升序排序:")
print(df.sort_values('工资'))
print("\n")

print("按工资降序排序:")
print(df.sort_values('工资', ascending=False))
print("\n")

print("按多列排序（先按评分降序，再按工资升序）:")
print(df.sort_values(['评分', '工资'], ascending=[False, True]))
print("\n")

# ===== 数据透视表 (Pivot Table) =====
sales_data = pd.DataFrame({
    '日期': ['2024-01-01', '2024-01-01', '2024-01-02', '2024-01-02'],
    '产品': ['A', 'B', 'A', 'B'],
    '销售额': [100, 150, 120, 130]
})

print("销售数据:")
print(sales_data)
print("\n")

pivot = pd.pivot_table(
    sales_data,
    values='销售额',
    index='日期',
    columns='产品',
    aggfunc='sum'
)
print("数据透视表:")
print(pivot)
print("\n")
```

# 统计与可视化

```python
# ===== 10. 统计与可视化 =====

import matplotlib.pyplot as plt

df = pd.DataFrame({
    '姓名': ['张三', '李四', '王五', '赵六', '孙七'],
    '年龄': [25, 30, 28, 22, 35],
    '工资': [8000, 12000, 10000, 7000, 15000],
    '评分': [85, 90, 80, 75, 92]
})

print("原始 DataFrame:")
print(df)
print("\n")

# ===== 统计函数 =====
print("统计摘要:")
print(df.describe())
print("\n")
# = 对 DataFrame 做快速统计分析 std = 标准差；25% = 第一四分位数(Q1)

print("均值:")
print(df[['年龄', '工资']].mean())
print("\n")

print("中位数:")
print(df[['年龄', '工资']].median())
print("\n")

print("标准差:")
print(df[['年龄', '工资']].std())
print("\n")

print("方差:")
print(df[['年龄', '工资']].var())
print("\n")

print("相关系数:")
print(df[['年龄', '工资', '评分']].corr())
print("\n")

# ===== 数据分布 =====
print("值计数（品种计数）:")
print(df['年龄'].value_counts())
print("\n")

# ===== 快速可视化 =====
print("生成可视化图表...\n")

# 相当于在pandas内部调用了matplotlib
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# 直方图
df['工资'].plot(kind='hist', ax=axes[0, 0], bins=5, title='工资分布')
axes[0, 0].set_xlabel('工资')

# 散点图
df.plot(x='年龄', y='工资', kind='scatter', ax=axes[0, 1], title='年龄 vs 工资')

# 箱线图
df['工资'].plot(kind='box', ax=axes[1, 0], title='工资箱线图')

# 折线图
df['评分'].plot(kind='line', ax=axes[1, 1], title='评分趋势')
axes[1, 1].set_xlabel('员工索引')

plt.tight_layout()
plt.show()

print("图表已显示")
print("\n")
```

# 常用操作速查表

## DataFrame 创建与查看

| 操作 | 代码 | 说明 |
|------|------|------|
| 从字典创建 | `pd.DataFrame({'A': [1,2], 'B': [3,4]})` | 最常用的方式 |
| 从列表创建 | `pd.DataFrame([[1,2], [3,4]], columns=['A','B'])` | 需要指定列名 |
| 从数组创建 | `pd.DataFrame(np.array([[1,2], [3,4]]))` | 使用 numpy 数组 |
| 读取 CSV | `pd.read_csv('file.csv')` | 指定 encoding 处理中文 |
| 读取 Excel | `pd.read_excel('file.xlsx', sheet_name='Sheet1')` | 指定工作表 |
| 查看前 n 行 | `df.head(n)` | 默认为 5 行 |
| 查看后 n 行 | `df.tail(n)` | 默认为 5 行 |
| 查看形状 | `df.shape` | 返回 (行数, 列数) |
| 查看列名 | `df.columns` | 返回列名列表 |
| 查看数据类型 | `df.dtypes` | 每列的数据类型 |
| 基本信息 | `df.info()` | 显示列、类型、非空计数 |
| 统计摘要 | `df.describe()` | 均值、标准差、四分位数等 |

## 数据选择与索引

| 操作 | 代码 | 说明 |
|------|------|------|
| 选择单列 | `df['列名']` 或 `df.列名` | 返回 Series |
| 选择多列 | `df[['列1', '列2']]` | 返回 DataFrame |
| 按位置选择行 | `df.iloc[0]` | 第一行 |
| 按位置选择行列 | `df.iloc[0:3, 1:3]` | 前 3 行，第 2-3 列 |
| 按标签选择行 | `df.loc[0]` | 索引为 0 的行 |
| 按标签选择行列 | `df.loc[0:2, ['A', 'B']]` | 0-2 行的 A、B 列 |
| 条件选择 | `df[df['工资'] > 10000]` | 工资 > 10000 的行 |
| 多条件AND | `df[(df['A'] > 5) & (df['B'] < 10)]` | 必须使用 & 而非 and |
| 多条件OR | `df[(df['A'] > 5) \| (df['B'] < 10)]` | 必须使用 \| 而非 or |
| 值在列表中 | `df[df['部门'].isin(['IT', 'HR'])]` | 在指定值中 |
| 字符串包含 | `df[df['名字'].str.contains('张')]` | 名字包含"张" |

## 数据清洗

| 操作 | 代码 | 说明 |
