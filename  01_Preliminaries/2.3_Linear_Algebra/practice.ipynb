# 2.3.6 降维

```python
x = torch.arange(4, dtype=torch.float32)
print(x, x.sum())
A = torch.arange(20, dtype=torch.float32).reshape(5, 4)
print(A)
print(A.sum(),A.shape)
```

```python
# axis=0 → 压缩"行" → 结果数量 = 列数  就是沿着行的方向拍扁
A_sum_axis0 = A.sum(axis=0)
print(A_sum_axis0, A_sum_axis0.shape)
# axis=1 → 压缩"列" → 结果数量 = 行数
A_sum_axis1 = A.sum(axis=1)
print(A_sum_axis1, A_sum_axis1.shape)
print(A.sum(axis=[0, 1])) # axis=[0, 1] → 压缩"行"和"列" → 结果数量 = 1 与 A.sum() 等价
print(A.mean(), A.sum() )
print(A.mean(axis=0), A.sum(axis=0)) # mean(axis=0) → 沿着行的方向拍扁，求平均值
```

```python
# 非降维求和（有时在求和的时候保持维度不变也很有必要）
print(A, A.shape, sep = "\n")
# 非降维
sum_A = A.sum(axis=1, keepdims=True)
print(sum_A, sum_A.shape,sep = "\n") # keepdims=True → 保持维度不变，结果数量 = 行数
# 普通
sum = A.sum(axis=1)
print(sum, sum.shape, sep = "\n")

print(A / sum_A) # A / sum_A → 广播机制：sum_A的列数为1，A的列数为4，sum_A会自动复制4份与A进行逐元素除法
```

```python
# 累积总和（就像银行账单的余额一样，每一行的结果都是前面所有行的和）
A.cumsum(axis=0)
# 结果行0 = 行0本身        = [ 0,  1,  2,  3]
# 结果行1 = 行0 + 行1      = [ 4,  6,  8, 10]
# 结果行2 = 行0+行1+行2    = [12, 15, 18, 21]
# 结果行3 = 前四行相加      = [24, 28, 32, 36]
# 结果行4 = 前五行相加      = [40, 45, 50, 55]
```
